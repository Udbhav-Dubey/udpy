```python
# Define the number of students and subjects
num_students = 10
subjects = ["Maths", "Science", "English", "IT"]
# Initialize dictionary to store marks
marks = {subject: [] for subject in subjects}
overall_marks = []
# Input marks for each student
print("Enter marks for each student (out of 100):")
for i in range(num_students):
    print(f"\nStudent {i+1}:")
    total = 0
    for subject in subjects:
    while True:
    try:
    mark = float(input(f" {subject}: "))
    if 0 <= mark <= 100:
break
else:
print(" Please enter a valid mark between 0 and 100.")
except ValueError:
print(" Invalid input. Please enter a number.")
marks[subject].append(mark)
total += mark
overall_marks.append(total)
Question 1: A class with 10 students wants to produce some information from the
results of the four standard tests in Maths, Science, English and IT. Each test is out
of 100 marks. The information output should be the highest, lowest and average mark
for each test and the highest, lowest and average mark overall. Write a program in
Python to complete this task.
Solution: # Define the number of students and subjects
num_students = 10
subjects = ["Maths", "Science", "English", "IT"]
# Initialize dictionary to store marks
marks = {subject: [] for subject in subjects}
overall_marks = []
# Input marks for each student
print("Enter marks for each student (out of 100):")
for i in range(num_students):
print(f"\nStudent {i+1}:")
total = 0
for subject in subjects:
while True:
try:
mark = float(input(f" {subject}: "))
if 0 <= mark <= 100:
break
else:
print(" Please enter a valid mark between 0 and 100.")
except ValueError:
print(" Invalid input. Please enter a number.")
marks[subject].append(mark)
total += mark
overall_marks.append(total)
# Calculate and display subject-wise stats
print("\nSubject-wise Statistics:")
for subject in subjects:
highest = max(marks[subject])
lowest = min(marks[subject])
average = sum(marks[subject]) / num_students
print(f"{subject}: Highest = {highest}, Lowest = {lowest}, Average = {average:.2f}")
# Calculate and display overall stats
overall_highest = max(overall_marks)
overall_lowest = min(overall_marks)
overall_average = sum(overall_marks) / num_students
print("\nOverall Statistics (Total of all 4 subjects per student):")
print(f"Highest Total Marks = {overall_highest}")
print(f"Lowest Total Marks = {overall_lowest}")
print(f"Average Total Marks = {overall_average:.2f}")
```
